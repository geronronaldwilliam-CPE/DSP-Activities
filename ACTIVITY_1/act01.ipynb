Original Sequence $x[n]$

```python
import numpy as np
import matplotlib.pyplot as plt

# Original Sequence Definition: x[n] = {2, -1, 3, 0, 4}
n_x = np.arange(0, 5)
x = np.array([2, -1, 3, 0, 4])

# Text Output
print("Original Sequence: x[n] = {2, -1, 3, 0, 4}")
print("\tn_x =", n_x.tolist())
print("\tx =", x.tolist())

# Graph Output
plt.figure(figsize=(8, 4))
plt.stem(n_x, x)
plt.title('Original Sequence x[n]')
plt.xlabel('n (sample index)')
plt.ylabel('x[n]')
plt.grid(True)
plt.axhline(0, color='black', lw=0.5)
plt.axvline(0, color='black', lw=0.5)
plt.xticks(np.arange(-5, 7, 1))
plt.tight_layout()
plt.show()
```

Time Shift $y_1[n] = x[n - 2]$

```python
# #1 Time Shift: y1[n] = x[n - 2]
n_y1 = n_x + 2
y1 = x.copy()

# Text Output
print("#1 Time Shift: y1[n] = x[n - 2]")
print("\tn_y1 =", n_y1.tolist())
print("\ty1 =", y1.tolist())

# Graph Output
plt.figure(figsize=(8, 4))
plt.stem(n_y1, y1)
plt.title('Time Shifted Sequence y1[n] = x[n - 2]')
plt.xlabel('n (sample index)')
plt.ylabel('y1[n]')
plt.grid(True)
plt.axhline(0, color='black', lw=0.5)
plt.axvline(0, color='black', lw=0.5)
plt.xticks(np.arange(-5, 7, 1))
plt.tight_layout()
plt.show()
```

Time Reversal $y_2[n] = x[-n]$

```python
# #2 Time Reversal: y2[n] = x[-n]
n_y2 = -n_x[::-1]
y2 = x[::-1]

# Text Output
print("#2 Time Reversal: y2[n] = x[-n]")
print("\tn_y2 =", n_y2.tolist())
print("\ty2 =", y2.tolist())

# Graph Output
plt.figure(figsize=(8, 4))
plt.stem(n_y2, y2)
plt.title('Time Reversed Sequence y2[n] = x[-n]')
plt.xlabel('n (sample index)')
plt.ylabel('y2[n]')
plt.grid(True)
plt.axhline(0, color='black', lw=0.5)
plt.axvline(0, color='black', lw=0.5)
plt.xticks(np.arange(-5, 7, 1))
plt.tight_layout()
plt.show()
```

Total Energy Computation

```python
# #3 Energy Computation
E = np.sum(np.abs(x) ** 2)

# Text Output
print("#3. Energy: E = sum |x[n]|^2")
print("\tEnergy E =", E)
```

Additional Graphs:

```python
n_elem = np.array([-4, -3, -2, -1, 0, 1, 2, 3, 4])

# 1. Impulse Response
xn_impulse = [0, 0, 0, 0, 1, 0, 0, 0, 0]
plt.figure(figsize=(7, 3))
plt.stem(n_elem, xn_impulse)
plt.title(r'Impulse Response $\delta[n]$')
plt.xlabel('Sampling Interval')
plt.ylabel('Sample')
plt.grid(True)
plt.show()

# 2. Step Response
xn_step = [0, 0, 0, 0, 1, 1, 1, 1, 1]
plt.figure(figsize=(7, 3))
plt.stem(n_elem, xn_step)
plt.title('Step Response $u[n]$')
plt.xlabel('Sampling Interval')
plt.ylabel('Sample')
plt.grid(True)
plt.show()

# 3. Ramp Response
xn_ramp = [-4, -3, -2, -1, 0, 1, 2, 3, 4]
plt.figure(figsize=(7, 3))
plt.stem(n_elem, xn_ramp)
plt.title('Ramp Response $r[n]$')
plt.xlabel('Sampling Interval')
plt.ylabel('Sample')
plt.grid(True)
plt.show()

# 4. Cosine Wave Response Wave
n_wave = np.arange(-6.2, 6.3, 0.1)
xn_wave = np.cos(n_wave)
plt.figure(figsize=(7, 3))
plt.stem(n_wave, xn_wave)
plt.title('Cosine Wave Response')
plt.xlabel('Sampling Interval')
plt.ylabel('Sample')
plt.grid(True)
plt.show()
```
